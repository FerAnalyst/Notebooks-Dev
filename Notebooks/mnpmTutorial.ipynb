# ¿Cómo instalar mnpm en tu equipo?
Nota: Se deberíaa tener instalado Node.js previamente antes de realizar los siguientes pasos:
1. Ingresa al siguiente enlace: https://pnpm.io/installation
2. Copia la siguiente línea de comando: npm install -g pnpm@latest-10
3. Abre el "cmd" o símbolo del sistema
4. Pega la líne de comando en el terminal y ejecuta.

5. Verifica la instalación usando el siguiente comando: pnpm -v
