# ¿Cómo crear y subir un repositorio en Github?
1. Guarda todos tus archivos en un único directorio



2. Crea un archivo llamado: **.gitignore** y agrega las siguientes lineas:

```js
.ipynb_checkpoints/
__pycache__/
*.pyc
.env
.DS_Store
Thumbs.db
```


3. Abrir la terminal usando **Ctrl + Ñ**



4. Usar los siguientes comandos:

```js
git init
git add .
git commit -m "Primer commit"
```
5. Crea un repositorio en Github



6. Guarda el enlace generado de repositorio: 
```js
https://github.com/FerAnalyst/Notebooks-Dev.git
```


7. Continua con los siguientes comandos:
```js
git remote add origin https://github.com/FerAnalyst/Notebooks-Dev.git
git branch -M main
git push -u origin main
```
